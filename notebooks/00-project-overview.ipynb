# Customer Lifetime Value (CLV) - Online Retail

## End-to-End Customer Analytics and CLV Modeling

### Project Overview

This project builds a complete CLV framework for an e-commerce retailer using the Online Retail dataset from Kaggle. The work follows the sequence of a commercial data science engagement: raw transaction data moves through cleaning, analysis, modeling, validation, and final outputs formatted for business intelligence tools.

The endpoint is a validated customer dataset and two CLV models, both designed to feed directly into marketing and CRM workflows - specifically retention targeting, customer prioritization, and budget allocation decisions.

### Business Context and Problem

Online retailers sit on large volumes of transaction data but rarely convert it into a forward-looking view of customer value. Marketing and CRM teams operate with questions that descriptive reporting alone cannot answer:

1. Which customers account for the majority of revenue?
2. Which customers will generate the most value over the next 6 to 12 months?
3. How should customers be grouped for retention offers versus upsell campaigns?
4. Where does a limited marketing budget produce the highest return?

This project answers these questions with a repeatable CLV and segmentation process. The outputs support both strategic planning and operational triggers such as churn-prevention offers and high-value upsell targeting.

### Notebook Structure and Analytical Flow

The analysis is organized into seven notebooks, each producing specific outputs that feed subsequent stages:

**Notebook 1: Data Cleaning and Preparation**

Resolves data quality issues systematically: negative quantities, canceled invoices, duplicate transactions, and missing customer identifiers. Produces the cleaned transaction dataset that serves as the single source of truth for all downstream work.

**Notebook 2: Exploratory Data Analysis**

Examines revenue distribution, customer concentration, seasonality, and purchasing patterns. Answers the question: Which customers account for the majority of revenue? Establishes the empirical basis for modeling decisions.

**Notebook 3: Customer 360 Table Construction**

Collapses each customer's history into one row with RFM metrics (Recency, Frequency, Monetary) plus behavioral features capturing purchasing intensity, tenure, and spend patterns. This table is the foundation for both CLV modeling and segmentation.

**Notebook 4: Probabilistic CLV (BG/NBD + Gamma-Gamma)**

Estimates repeat purchase behavior and expected monetary value. Produces 12-month forward-looking customer contribution. Supports long-term customer prioritization and retention planning.

**Notebook 5: Machine Learning CLV (6-Month Horizon)**

Predicts customer revenue over 6 months using supervised learning. Features use a strict point-in-time cutoff, no information from after the prediction date enters training. This aligns with marketing planning cycles where near-term revenue impact drives decisions.

**Notebook 6: Customer Segmentation**

Segments customers using RFM-based scoring followed by K-Means clustering. Translates each segment into plain-language profiles with specific marketing and CRM actions assigned.

**Notebook 7: Targeting Uplift Simulation and Export**

Quantifies the incremental revenue from targeting top X% of customers by predicted CLV versus random selection. Exports all final outputs as fact and dimension tables with a data dictionary for Power BI integration.

### CLV Definitions and Modeling Strategy

**12-Month CLV (Probabilistic):** Expected customer contribution over 12 months, modeled from historical frequency and monetary value patterns. Used for long-term value estimation and strategic decisions.

**6-Month CLV (Machine Learning):** Predicted customer revenue over 6 months, using supervised learning with temporal validation. Used for short-term campaign planning and performance measurement.

Running both models is deliberate. The probabilistic approach offers stability and interpretability. The ML approach captures non-linear patterns and aligns with shorter planning windows. Together they cover the two timelines on which marketing decisions actually get made.

### Assumptions and Risk Controls

**Returns and Cancellations:** Negative quantities and canceled invoices are identified and quarantined during cleaning. Revenue metrics are computed net of these transactions where appropriate.

**Missing Customer Identifiers:** Transactions without a valid CustomerID are excluded from customer-level modeling. They are retained in aggregate revenue summaries for top-line context.

**Data Leakage Prevention:** All ML features are constructed from data available on or before the cutoff date. No future information enters training or evaluation sets. This ensures model performance estimates reflect real-world deployment conditions.
